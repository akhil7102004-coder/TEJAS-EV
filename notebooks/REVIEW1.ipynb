```python
print("Dataset Shape:")
print(df.shape)

print("\nNumber of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])

print("\nColumn Names:")
for i, col in enumerate(df.columns, 1):
    print(f"{i}. {col}")
```

```python
print("DATA TYPES")


print(df.dtypes)

print("\n\nMISSING VALUES")


missing = df.isnull().sum()

missing_table = pd.DataFrame({
    "Missing Count": missing,
    "Missing %": (missing / len(df) * 100).round(2)
})

print(missing_table)
```

```python
print("DUPLICATE CHECK")


duplicate_rows = df.duplicated().sum()

print("Duplicate rows:", duplicate_rows)


print("\nDEPOT CHECK")


print("Unique Depots:", df["Depot ID"].nunique())
print("Unique Depot Names:", df["Depot Name"].nunique())

print("\nFirst 15 Depots:")
print(df[["Depot ID", "Depot Name", "District"]]
      .drop_duplicates()
      .head(15)
      .to_string(index=False))


print("\nDISTRICT CHECK")


print("Unique Districts:", df["District"].nunique())
print(sorted(df["District"].unique()))


print("\nMONTH CHECK")


print("Unique Months:", df["Month"].nunique())
print(sorted(df["Month"].unique()))


print("\nFISCAL YEAR CHECK")


print("Unique Fiscal Years:", df["Fiscal Year"].nunique())
print(sorted(df["Fiscal Year"].unique()))


print("\nDATA STATUS")


print(df["Data Status"].value_counts())


print("\nDEPOT OBSERVATION STATUS")

print(df["Depot Observation Status"].value_counts())


print("\nDATA PROVENANCE")

print(df["Data Provenance"].value_counts())
```

```python
numeric_cols = df.select_dtypes(include=["int64", "float64"]).columns

print("NUMERICAL RANGE CHECK")

for col in numeric_cols:
    print(f"\n{col}")
    print(f"  Min    : {df[col].min():,.2f}")
    print(f"  Max    : {df[col].max():,.2f}")
    print(f"  Mean   : {df[col].mean():,.2f}")
    print(f"  Median : {df[col].median():,.2f}")


# ------------------------------------------------------------
# Check negative values
# ------------------------------------------------------------

print("\n\nNEGATIVE VALUE CHECK")

negative_counts = (df[numeric_cols] < 0).sum()

for col, count in negative_counts.items():
    print(f"{col}: {count}")


# ------------------------------------------------------------
# Check zero values
# ------------------------------------------------------------

print("\n\nZERO VALUE CHECK")

zero_counts = (df[numeric_cols] == 0).sum()

for col, count in zero_counts.items():
    print(f"{col}: {count}")
```

```python
print("DERIVED FEATURE CONSISTENCY CHECK")


# ------------------------------------------------------------
# 1. Passengers per KM
# ------------------------------------------------------------

expected_ppkm = df["Passengers"] / df["Effective KM"]

ppkm_error = (
    df["Passengers per KM"] - expected_ppkm
).abs()

print("\n1. PASSENGERS PER KM")
print("Maximum calculation error:",
      ppkm_error.max())

print("Mean calculation error:",
      ppkm_error.mean())


# ------------------------------------------------------------
# 2. Diesel consumption
# ------------------------------------------------------------

diesel_rate = (
    df["Estimated Diesel Litres"] /
    df["Effective KM"]
)

print("\n2. DIESEL CONSUMPTION")
print("Implied litres per KM:")
print("Min   :", diesel_rate.min())
print("Max   :", diesel_rate.max())
print("Mean  :", diesel_rate.mean())
print("Median:", diesel_rate.median())


# ------------------------------------------------------------
# 3. CO2 per litre of diesel
# ------------------------------------------------------------

co2_per_litre = (
    df["Estimated CO2 (Tonnes)"] /
    df["Estimated Diesel Litres"]
)

print("\n3. CO2 EMISSION FACTOR")
print("Implied tonnes CO2 per litre:")
print("Min   :", co2_per_litre.min())
print("Max   :", co2_per_litre.max())
print("Mean  :", co2_per_litre.mean())
print("Median:", co2_per_litre.median())


# ------------------------------------------------------------
# 4. EV energy consumption
# ------------------------------------------------------------

ev_energy_rate = (
    df["Estimated EV Energy (MWh)"] * 1000 /
    df["Effective KM"]
)

print("\n4. EV ENERGY CONSUMPTION")
print("Implied kWh per KM:")
print("Min   :", ev_energy_rate.min())
print("Max   :", ev_energy_rate.max())
print("Mean  :", ev_energy_rate.mean())
print("Median:", ev_energy_rate.median())


# ------------------------------------------------------------
# 5. OPEX saving
# ------------------------------------------------------------

expected_saving = (
    df["Effective KM"] *
    (
        df["Diesel Cost (INR/KM)"] -
        df["EV Cost (INR/KM)"]
    )
)

saving_error = (
    df["Potential EV OPEX Saving (INR)"] -
    expected_saving
).abs()

print("\n5. EV OPEX SAVING")
print("Maximum calculation error:",
      saving_error.max())

print("Mean calculation error:",
      saving_error.mean())


# ------------------------------------------------------------
# 6. EV suitability score range
# ------------------------------------------------------------

print("\n6. EV SUITABILITY SCORE")

print("Minimum:",
      df["EV Suitability Score"].min())

print("Maximum:",
      df["EV Suitability Score"].max())

print("Values outside 0-1:",
      (
          (df["EV Suitability Score"] < 0) |
          (df["EV Suitability Score"] > 1)
      ).sum())


# ------------------------------------------------------------
# FINAL SUMMARY
# ------------------------------------------------------------
```

```python
# Calculate percentile ranks
depot_ev["Suitability Percentile"] = (
    depot_ev["EV Suitability Score"].rank(pct=True)
)

depot_ev["KM Percentile"] = (
    depot_ev["Effective KM"].rank(pct=True)
)

depot_ev["Savings Percentile"] = (
    depot_ev["Potential EV OPEX Saving (INR)"].rank(pct=True)
)

# Combined screening indicator
depot_ev["EV Opportunity Index"] = (
    depot_ev["Suitability Percentile"] * 0.50
    + depot_ev["KM Percentile"] * 0.25
    + depot_ev["Savings Percentile"] * 0.25
)

# Categorize
depot_ev["EV Opportunity Category"] = pd.cut(
    depot_ev["EV Opportunity Index"],
    bins=[0, 0.33, 0.66, 1.0],
    labels=["Lower Opportunity", "Moderate Opportunity", "Higher Opportunity"],
    include_lowest=True
)

# Sort
opportunity = depot_ev.sort_values(
    "EV Opportunity Index",
    ascending=False
).reset_index(drop=True)

print("EV OPPORTUNITY SCREENING")
print("=" * 110)

display(
    opportunity[
        [
            "Depot ID",
            "Depot Name",
            "District",
            "EV Suitability Score",
            "Effective KM",
            "Potential EV OPEX Saving (INR)",
            "EV Opportunity Index",
            "EV Opportunity Category"
        ]
    ].head(20).round(3)
)
```

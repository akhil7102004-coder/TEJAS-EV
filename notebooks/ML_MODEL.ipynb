```python
import pandas as pd
import numpy as np

# Load the KSRTC master dataset
df = pd.read_csv("../data/raw/KSRTC_EV_Master_Dataset.csv")

print("Dataset Shape:", df.shape)

print("\nColumns:")
for col in df.columns:
    print(col)

print("\nFirst 5 rows:")
display(df.head())
```

```python
print("Missing values:")
display(df.isnull().sum())

print("\nData types:")
display(df.dtypes)
```

```python
print("Number of depots:", df["Depot_ID"].nunique())
print("Number of districts:", df["District"].nunique())

print("\nDepots:")
print(sorted(df["Depot_ID"].unique()))
```

```python
# ============================================
# STEP 6: CREATE DEPOT-LEVEL ML DATASET
# ============================================

# Create passengers per schedule
df["Passengers_per_Schedule"] = (
    df["Passengers"] / df["Schedules Allocated"]
).replace([np.inf, -np.inf], np.nan).fillna(0)


# Aggregate monthly data to depot level
depot_ml = (
    df.groupby(
        [
            "Depot_ID",
            "Depot_Name",
            "District",
            "Terrain_Class",
            "Terrain_Score"
        ],
        as_index=False
    )
    .agg(
        # -------------------------
        # OPERATIONAL FEATURES
        # -------------------------
        Avg_Buses=("Buses Allocated", "mean"),
        Avg_Schedules=("Schedules Allocated", "mean"),
        Avg_Effective_KM=("Effective KM", "mean"),

        # -------------------------
        # PASSENGER DEMAND
        # -------------------------
        Avg_Passengers=("Passengers", "mean"),
        Avg_Passengers_per_Bus=("Passengers_per_Bus", "mean"),
        Avg_Passengers_per_Schedule=("Passengers_per_Schedule", "mean"),

        # -------------------------
        # DIESEL / ENVIRONMENT
        # -------------------------
        Avg_Diesel_Litres=("Estimated Diesel Litres", "mean"),
        Avg_CO2_Tonnes=("Estimated CO2 (Tonnes)", "mean"),

        # -------------------------
        # EV / ENERGY
        # -------------------------
        Avg_EV_Energy_MWh=("Estimated EV Energy (MWh)", "mean"),

        # -------------------------
        # ECONOMIC
        # -------------------------
        Avg_OPEX_Saving_INR=("Potential EV OPEX Saving (INR)", "mean"),

        # -------------------------
        # TOTAL IMPACT
        # -------------------------
        Total_Effective_KM=("Effective KM", "sum"),
        Total_Diesel_Litres=("Estimated Diesel Litres", "sum"),
        Total_CO2_Tonnes=("Estimated CO2 (Tonnes)", "sum"),
        Total_EV_Energy_MWh=("Estimated EV Energy (MWh)", "sum"),
        Total_OPEX_Saving_INR=("Potential EV OPEX Saving (INR)", "sum"),

        # Number of months available
        Months_Observed=("Month_Number", "count")
    )
)


# ============================================
# ANNUALIZED IMPACT
# ============================================

depot_ml["Annualized_Effective_KM"] = (
    depot_ml["Total_Effective_KM"]
    / depot_ml["Months_Observed"] * 12
)

depot_ml["Annualized_CO2_Reduction_Tonnes"] = (
    depot_ml["Total_CO2_Tonnes"]
    / depot_ml["Months_Observed"] * 12
)

depot_ml["Annualized_OPEX_Saving_INR"] = (
    depot_ml["Total_OPEX_Saving_INR"]
    / depot_ml["Months_Observed"] * 12
)


# ============================================
# CHECK RESULT
# ============================================

print("Original dataset:", df.shape)
print("Depot-level dataset:", depot_ml.shape)

print("\nNumber of unique depots:",
      depot_ml["Depot_ID"].nunique())

display(depot_ml.head())
```

```python
# ============================================
# STEP 6: CREATE FULL MONTHLY ML DATASET
# ============================================

# Work from the original 5,520-row dataset
monthly_ml = df.copy()

# Create passengers per schedule
monthly_ml["Passengers_per_Schedule"] = (
    monthly_ml["Passengers"] /
    monthly_ml["Schedules Allocated"]
).replace([np.inf, -np.inf], np.nan).fillna(0)

print("Monthly ML dataset shape:", monthly_ml.shape)

display(monthly_ml.head())
```

```python
# ============================================
# STEP 7: CREATE NORMALIZED ANALYTICAL SCORES
# ============================================

from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

score_columns = [
    "Effective KM",
    "Passengers",
    "Passengers_per_Bus",
    "Passengers_per_Schedule",
    "Estimated Diesel Litres",
    "Estimated CO2 (Tonnes)",
    "Potential EV OPEX Saving (INR)"
]

scaled_values = scaler.fit_transform(
    monthly_ml[score_columns]
)

scaled_df = pd.DataFrame(
    scaled_values,
    columns=[
        "KM_Score",
        "Passenger_Score",
        "Passenger_Bus_Score",
        "Passenger_Schedule_Score",
        "Diesel_Score",
        "CO2_Score",
        "OPEX_Saving_Score"
    ],
    index=monthly_ml.index
)

monthly_ml = pd.concat(
    [monthly_ml, scaled_df],
    axis=1
)

display(
    monthly_ml[
        [
            "Depot_ID",
            "Year",
            "Month",
            "KM_Score",
            "Passenger_Score",
            "Diesel_Score",
            "CO2_Score",
            "OPEX_Saving_Score",
            "Terrain_Score"
        ]
    ].head()
)
```

```python
# ============================================
# STEP 8: EXPERT-GUIDED EV TRANSITION SCORE
# ============================================

monthly_ml["EV_Transition_Score"] = (
    0.25 * monthly_ml["KM_Score"]
    + 0.20 * monthly_ml["Passenger_Score"]
    + 0.10 * monthly_ml["Passenger_Bus_Score"]
    + 0.10 * monthly_ml["Passenger_Schedule_Score"]
    + 0.10 * monthly_ml["Diesel_Score"]
    + 0.10 * monthly_ml["OPEX_Saving_Score"]
    + 0.05 * monthly_ml["CO2_Score"]
    + 0.10 * monthly_ml["Terrain_Score"]
)

monthly_ml["EV_Transition_Score"] = (
    monthly_ml["EV_Transition_Score"].clip(0, 1)
)

display(
    monthly_ml[
        [
            "Depot_ID",
            "Depot_Name",
            "EV_Transition_Score",
            "Terrain_Class"
        ]
    ].head(10)
)
```

```python
# ============================================
# STEP 9: INSPECT SCORE DISTRIBUTION
# ============================================

print(monthly_ml["EV_Transition_Score"].describe())

import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))

plt.hist(
    monthly_ml["EV_Transition_Score"],
    bins=20
)

plt.xlabel("EV Transition Score")
plt.ylabel("Number of Observations")
plt.title("Distribution of EV Transition Scores")

plt.show()
```

```python
# ============================================
# STEP 10: CREATE EV SUITABILITY COMPONENTS
# ============================================

# Percentile-based normalization
# This reduces the influence of extreme outliers.

monthly_ml["KM_Suitability"] = (
    monthly_ml["Effective KM"].rank(pct=True)
)

monthly_ml["Bus_Suitability"] = (
    monthly_ml["Buses Allocated"].rank(pct=True)
)

monthly_ml["Schedule_Suitability"] = (
    monthly_ml["Schedules Allocated"].rank(pct=True)
)

monthly_ml["Passenger_Suitability"] = (
    monthly_ml["Passengers"].rank(pct=True)
)

monthly_ml["Passenger_Bus_Suitability"] = (
    monthly_ml["Passengers_per_Bus"].rank(pct=True)
)

monthly_ml["Passenger_Schedule_Suitability"] = (
    monthly_ml["Passengers_per_Schedule"].rank(pct=True)
)

# EV energy requirement is NOT directly "good" or "bad".
# We keep it as an operational planning indicator.
monthly_ml["Energy_Intensity_Score"] = (
    monthly_ml["Estimated EV Energy (MWh)"].rank(pct=True)
)

display(
    monthly_ml[
        [
            "Depot_ID",
            "KM_Suitability",
            "Bus_Suitability",
            "Schedule_Suitability",
            "Passenger_Suitability",
            "Passenger_Bus_Suitability",
            "Passenger_Schedule_Suitability",
            "Terrain_Score"
        ]
    ].head(10)
)
```

```python
# ============================================
# STEP 11: EV SUITABILITY SCORE
# ============================================

monthly_ml["EV_Suitability_Score"] = (
    0.30 * monthly_ml["KM_Suitability"]
    + 0.15 * monthly_ml["Passenger_Suitability"]
    + 0.10 * monthly_ml["Passenger_Bus_Suitability"]
    + 0.10 * monthly_ml["Passenger_Schedule_Suitability"]
    + 0.10 * monthly_ml["Bus_Suitability"]
    + 0.05 * monthly_ml["Schedule_Suitability"]
    + 0.20 * monthly_ml["Terrain_Score"]
)

monthly_ml["EV_Suitability_Score"] = (
    monthly_ml["EV_Suitability_Score"].clip(0, 1)
)

print(
    monthly_ml["EV_Suitability_Score"].describe()
)

display(
    monthly_ml[
        [
            "Depot_ID",
            "Depot_Name",
            "Terrain_Class",
            "Terrain_Score",
            "EV_Suitability_Score"
        ]
    ].head(10)
)
```

```python
# ============================================
# STEP 15: CREATE 3-CLASS EV TRANSITION LABEL
# ============================================

def assign_ev_category(row):
    
    score = row["EV_Suitability_Score"]
    terrain = row["Terrain_Class"]
    
    # -----------------------------------------
    # TERRAIN CONSTRAINTS
    # -----------------------------------------
    
    # Steep terrain is not recommended for EV
    if terrain == "Steep":
        return "Diesel Preferred"
    
    # Hilly terrain can be considered, but not
    # classified as a strong EV candidate
    if terrain == "Hilly":
        if score >= 0.55:
            return "Conditional"
        else:
            return "Diesel Preferred"
    
    # -----------------------------------------
    # GENERAL SUITABILITY
    # -----------------------------------------
    
    if score >= 0.67:
        return "EV Suitable"
    
    elif score >= 0.43:
        return "Conditional"
    
    else:
        return "Diesel Preferred"


monthly_ml["EV_Transition_Category"] = (
    monthly_ml.apply(assign_ev_category, axis=1)
)

# Check class distribution
print("EV Transition Category Distribution:")
display(
    monthly_ml["EV_Transition_Category"]
    .value_counts()
)

print("\nPercentage distribution:")
display(
    monthly_ml["EV_Transition_Category"]
    .value_counts(normalize=True).mul(100).round(2)
)
```

```python
# ============================================
# STEP 17: INSPECT GENERATED CATEGORIES
# ============================================

category_summary = (
    monthly_ml
    .groupby("EV_Transition_Category")
    .agg(
        Observations=("Depot_ID", "count"),
        Avg_Suitability=("EV_Suitability_Score", "mean"),
        Avg_KM=("Effective KM", "mean"),
        Avg_Passengers=("Passengers", "mean"),
        Avg_CO2=("Estimated CO2 (Tonnes)", "mean"),
        Avg_OPEX_Saving=("Potential EV OPEX Saving (INR)", "mean"),
        Avg_Terrain_Score=("Terrain_Score", "mean")
    )
    .sort_values("Avg_Suitability", ascending=False)
)

display(category_summary)
```

```python
# ============================================
# STEP 18: TERRAIN VS EV CATEGORY
# ============================================

terrain_category = pd.crosstab(
    monthly_ml["Terrain_Class"],
    monthly_ml["EV_Transition_Category"],
    normalize="index"
).mul(100).round(2)

display(terrain_category)
```

```python
# ============================================
# STEP 19: OPERATIONAL & DEMAND SCORES
# ============================================

# Operational intensity
monthly_ml["Operational_Intensity_Score"] = (
    0.55 * monthly_ml["KM_Suitability"]
    + 0.25 * monthly_ml["Bus_Suitability"]
    + 0.20 * monthly_ml["Schedule_Suitability"]
)

# Passenger demand
monthly_ml["Passenger_Demand_Score"] = (
    0.40 * monthly_ml["Passenger_Suitability"]
    + 0.30 * monthly_ml["Passenger_Bus_Suitability"]
    + 0.30 * monthly_ml["Passenger_Schedule_Suitability"]
)

display(
    monthly_ml[
        [
            "Depot_ID",
            "Depot_Name",
            "Operational_Intensity_Score",
            "Passenger_Demand_Score",
            "Terrain_Score"
        ]
    ].head(10)
)
```

```python
# ============================================
# STEP 20: REFINED EV SUITABILITY SCORE
# ============================================

monthly_ml["EV_Suitability_Score"] = (
    0.50 * monthly_ml["Operational_Intensity_Score"]
    + 0.30 * monthly_ml["Passenger_Demand_Score"]
    + 0.20 * monthly_ml["Terrain_Score"]
)

monthly_ml["EV_Suitability_Score"] = (
    monthly_ml["EV_Suitability_Score"].clip(0, 1)
)

print(monthly_ml["EV_Suitability_Score"].describe())

display(
    monthly_ml[
        [
            "Depot_ID",
            "Depot_Name",
            "Terrain_Class",
            "Operational_Intensity_Score",
            "Passenger_Demand_Score",
            "Terrain_Score",
            "EV_Suitability_Score"
        ]
    ].head(10)
)
```

```python
# ============================================
# STEP 21: FINAL EV TRANSITION CATEGORIES
# ============================================

def assign_ev_category(row):

    score = row["EV_Suitability_Score"]
    terrain = row["Terrain_Class"]

    # -----------------------------------------
    # HARD TERRAIN CONSTRAINT
    # -----------------------------------------
    
    # Steep terrain → Diesel Preferred
    if terrain == "Steep":
        return "Diesel Preferred"

    # Hilly terrain → maximum Conditional
    if terrain == "Hilly":
        if score >= 0.43:
            return "Conditional"
        else:
            return "Diesel Preferred"

    # -----------------------------------------
    # GENERAL SCORE-BASED CLASSIFICATION
    # -----------------------------------------
    
    if score >= 0.67:
        return "EV Suitable"

    elif score >= 0.43:
        return "Conditional"

    else:
        return "Diesel Preferred"


# Generate FINAL labels
monthly_ml["EV_Transition_Category"] = (
    monthly_ml.apply(assign_ev_category, axis=1)
)


# -----------------------------------------
# CLASS DISTRIBUTION
# -----------------------------------------

print("FINAL EV TRANSITION CATEGORY DISTRIBUTION")
print("=" * 50)

category_counts = (
    monthly_ml["EV_Transition_Category"]
    .value_counts()
)

display(category_counts)


print("\nPercentage Distribution")
print("=" * 50)

category_percent = (
    monthly_ml["EV_Transition_Category"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

display(category_percent)
```

```python
# ============================================
# STEP 22: VALIDATE FINAL CATEGORIES
# ============================================

category_summary = (
    monthly_ml
    .groupby("EV_Transition_Category")
    .agg(
        Observations=("Depot_ID", "count"),
        Avg_Suitability=("EV_Suitability_Score", "mean"),
        Avg_Operational=("Operational_Intensity_Score", "mean"),
        Avg_Demand=("Passenger_Demand_Score", "mean"),
        Avg_KM=("Effective KM", "mean"),
        Avg_Passengers=("Passengers", "mean"),
        Avg_CO2=("Estimated CO2 (Tonnes)", "mean"),
        Avg_OPEX_Saving=("Potential EV OPEX Saving (INR)", "mean"),
        Avg_Terrain=("Terrain_Score", "mean")
    )
    .sort_values("Avg_Suitability", ascending=False)
)

display(category_summary)
```

```python
# ============================================
# STEP 23: OPERATIONAL INTENSITY VS CATEGORY
# ============================================

operational_summary = (
    monthly_ml
    .groupby("EV_Transition_Category")
    ["Operational_Intensity_Score"]
    .agg(["mean", "min", "max"])
    .round(3)
)

display(operational_summary)
```

```python
# ============================================
# STEP 24: PASSENGER DEMAND VS CATEGORY
# ============================================

demand_summary = (
    monthly_ml
    .groupby("EV_Transition_Category")
    ["Passenger_Demand_Score"]
    .agg(["mean", "min", "max"])
    .round(3)
)

display(demand_summary)
```

```python
# ============================================
# STEP 25: DEFINE ML FEATURES
# ============================================

# Raw features used by the ML model
feature_columns = [
    "Buses Allocated",
    "Schedules Allocated",
    "Passengers",
    "Estimated Diesel Litres",
    "Estimated CO2 (Tonnes)",
    "Effective KM",
    "Estimated EV Energy (MWh)",
    "Passengers_per_Bus",
    "Passengers_per_Schedule",
    "Diesel_Litres_per_Bus",
    "Diesel_Litres_per_Schedule",
    "CO2_per_Bus_Tonnes",
    "Terrain_Score"
]

target_column = "EV_Transition_Category"

X = monthly_ml[feature_columns].copy()
y = monthly_ml[target_column].copy()

print("Number of features:", len(feature_columns))
print("\nFeatures:")
for feature in feature_columns:
    print("-", feature)

print("\nX shape:", X.shape)
print("y shape:", y.shape)
```

```python
# ============================================
# STEP 26: GROUP-AWARE TRAIN/TEST SPLIT
# ============================================

from sklearn.model_selection import GroupShuffleSplit

groups = monthly_ml["Depot_ID"]

gss = GroupShuffleSplit(
    n_splits=1,
    test_size=0.20,
    random_state=42
)

train_idx, test_idx = next(
    gss.split(X, y, groups=groups)
)

X_train = X.iloc[train_idx].copy()
X_test = X.iloc[test_idx].copy()

y_train = y.iloc[train_idx].copy()
y_test = y.iloc[test_idx].copy()

groups_train = groups.iloc[train_idx]
groups_test = groups.iloc[test_idx]

print("Training observations:", len(X_train))
print("Testing observations:", len(X_test))

print("\nTraining depots:", groups_train.nunique())
print("Testing depots:", groups_test.nunique())

print("\nTraining classes:")
display(y_train.value_counts())

print("\nTesting classes:")
display(y_test.value_counts())
```

```python
# ============================================
# STEP 27: PREPROCESSING
# ============================================

from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

preprocessor = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

print("Preprocessing pipeline created.")
```

```python
# ============================================
# STEP 28: LOGISTIC REGRESSION
# ============================================

from sklearn.linear_model import LogisticRegression

logistic_model = Pipeline([
    ("preprocessor", preprocessor),
    (
        "classifier",
        LogisticRegression(
            max_iter=2000,
            class_weight="balanced",
            random_state=42
        )
    )
])

logistic_model.fit(X_train, y_train)

logistic_pred = logistic_model.predict(X_test)

print("Logistic Regression trained successfully.")
```

```python
# ============================================
# STEP 29: LOGISTIC REGRESSION EVALUATION
# ============================================

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

logistic_accuracy = accuracy_score(
    y_test,
    logistic_pred
)

logistic_precision = precision_score(
    y_test,
    logistic_pred,
    average="macro",
    zero_division=0
)

logistic_recall = recall_score(
    y_test,
    logistic_pred,
    average="macro",
    zero_division=0
)

logistic_f1 = f1_score(
    y_test,
    logistic_pred,
    average="macro",
    zero_division=0
)

print("LOGISTIC REGRESSION")
print("=" * 40)

print("Accuracy :", round(logistic_accuracy, 4))
print("Precision:", round(logistic_precision, 4))
print("Recall   :", round(logistic_recall, 4))
print("Macro F1 :", round(logistic_f1, 4))

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        logistic_pred,
        zero_division=0
    )
)
```

```python
# ============================================
# STEP 30: CONFUSION MATRIX
# ============================================

import matplotlib.pyplot as plt
from sklearn.metrics import ConfusionMatrixDisplay

ConfusionMatrixDisplay.from_predictions(
    y_test,
    logistic_pred,
    xticks_rotation=45
)

plt.title("Logistic Regression - Confusion Matrix")
plt.tight_layout()
plt.show()
```

```python
# ============================================
# STEP 31: DECISION TREE
# ============================================

from sklearn.tree import DecisionTreeClassifier

decision_tree_model = Pipeline([
    ("preprocessor", preprocessor),
    (
        "classifier",
        DecisionTreeClassifier(
            max_depth=6,
            min_samples_leaf=10,
            class_weight="balanced",
            random_state=42
        )
    )
])

decision_tree_model.fit(X_train, y_train)

dt_pred = decision_tree_model.predict(X_test)

print("Decision Tree trained successfully.")
```

```python
# ============================================
# STEP 32: DECISION TREE EVALUATION
# ============================================

dt_accuracy = accuracy_score(y_test, dt_pred)

dt_precision = precision_score(
    y_test,
    dt_pred,
    average="macro",
    zero_division=0
)

dt_recall = recall_score(
    y_test,
    dt_pred,
    average="macro",
    zero_division=0
)

dt_f1 = f1_score(
    y_test,
    dt_pred,
    average="macro",
    zero_division=0
)

print("DECISION TREE")
print("=" * 40)

print("Accuracy :", round(dt_accuracy, 4))
print("Precision:", round(dt_precision, 4))
print("Recall   :", round(dt_recall, 4))
print("Macro F1 :", round(dt_f1, 4))

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        dt_pred,
        zero_division=0
    )
)
```

```python
# ============================================
# STEP 33: RANDOM FOREST
# ============================================

from sklearn.ensemble import RandomForestClassifier

random_forest_model = Pipeline([
    ("preprocessor", preprocessor),
    (
        "classifier",
        RandomForestClassifier(
            n_estimators=300,
            max_depth=10,
            min_samples_leaf=5,
            class_weight="balanced",
            random_state=42,
            n_jobs=-1
        )
    )
])

random_forest_model.fit(X_train, y_train)

rf_pred = random_forest_model.predict(X_test)

print("Random Forest trained successfully.")
```

```python
# ============================================
# STEP 34: RANDOM FOREST EVALUATION
# ============================================

rf_accuracy = accuracy_score(y_test, rf_pred)

rf_precision = precision_score(
    y_test,
    rf_pred,
    average="macro",
    zero_division=0
)

rf_recall = recall_score(
    y_test,
    rf_pred,
    average="macro",
    zero_division=0
)

rf_f1 = f1_score(
    y_test,
    rf_pred,
    average="macro",
    zero_division=0
)

print("RANDOM FOREST")
print("=" * 40)

print("Accuracy :", round(rf_accuracy, 4))
print("Precision:", round(rf_precision, 4))
print("Recall   :", round(rf_recall, 4))
print("Macro F1 :", round(rf_f1, 4))

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        rf_pred,
        zero_division=0
    )
)
```

```python
# ============================================
# STEP 35: MODEL COMPARISON
# ============================================

model_comparison = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "Decision Tree",
        "Random Forest"
    ],
    "Accuracy": [
        logistic_accuracy,
        dt_accuracy,
        rf_accuracy
    ],
    "Macro Precision": [
        logistic_precision,
        dt_precision,
        rf_precision
    ],
    "Macro Recall": [
        logistic_recall,
        dt_recall,
        rf_recall
    ],
    "Macro F1": [
        logistic_f1,
        dt_f1,
        rf_f1
    ]
})

model_comparison = (
    model_comparison
    .sort_values("Macro F1", ascending=False)
    .reset_index(drop=True)
)

display(
    model_comparison.round(4)
)
```

```python
# ============================================
# STEP 39: DECISION TREE HYPERPARAMETER TUNING
# ============================================

from sklearn.model_selection import GridSearchCV, GroupKFold
from sklearn.tree import DecisionTreeClassifier

# Fresh pipeline
dt_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    (
        "classifier",
        DecisionTreeClassifier(
            class_weight="balanced",
            random_state=42
        )
    )
])

# Hyperparameter grid
dt_param_grid = {
    "classifier__max_depth": [3, 4, 5, 6, 8, 10, 12, None],
    "classifier__min_samples_split": [2, 5, 10, 20],
    "classifier__min_samples_leaf": [1, 5, 10, 20],
    "classifier__criterion": ["gini", "entropy"]
}

# Group-aware cross-validation
group_kfold = GroupKFold(n_splits=5)

dt_grid = GridSearchCV(
    estimator=dt_pipeline,
    param_grid=dt_param_grid,
    scoring="f1_macro",
    cv=group_kfold,
    n_jobs=-1,
    verbose=1
)

dt_grid.fit(
    X_train,
    y_train,
    groups=groups_train
)

print("Decision Tree tuning completed.")
print("\nBest parameters:")
print(dt_grid.best_params_)

print("\nBest cross-validation Macro F1:")
print(round(dt_grid.best_score_, 4))
```

```python
# ============================================================
# FINAL EV TRANSITION PRIORITY SCORE
# ============================================================

import pandas as pd
import numpy as np

priority_df = final_depot.copy()


# ------------------------------------------------------------
# 1. Normalize important factors using percentile ranking
# ------------------------------------------------------------

priority_df["Suitability_Rank"] = (
    priority_df["Avg_EV_Suitability"].rank(pct=True)
)

priority_df["OPEX_Rank"] = (
    priority_df["Total_OPEX_Saving_INR"].rank(pct=True)
)

priority_df["Passenger_Rank"] = (
    priority_df["Avg_Passengers"].rank(pct=True)
)

priority_df["KM_Rank"] = (
    priority_df["Avg_Effective_KM"].rank(pct=True)
)

priority_df["Operational_Rank"] = (
    priority_df["Avg_Operational_Intensity"].rank(pct=True)
)


# ------------------------------------------------------------
# 2. Calculate final transition priority score
# ------------------------------------------------------------

priority_df["EV_Transition_Priority_Score"] = (
    0.30 * priority_df["Suitability_Rank"]
    + 0.25 * priority_df["OPEX_Rank"]
    + 0.20 * priority_df["Passenger_Rank"]
    + 0.15 * priority_df["KM_Rank"]
    + 0.10 * priority_df["Operational_Rank"]
)


# ------------------------------------------------------------
# 3. Rank depots
# ------------------------------------------------------------

priority_df["Final_Priority_Rank"] = (
    priority_df["EV_Transition_Priority_Score"]
    .rank(ascending=False, method="min")
    .astype(int)
)


# ------------------------------------------------------------
# 4. Sort
# ------------------------------------------------------------

priority_df = priority_df.sort_values(
    "EV_Transition_Priority_Score",
    ascending=False
).reset_index(drop=True)


# ------------------------------------------------------------
# 5. Show Top 10
# ------------------------------------------------------------

top10 = priority_df.head(10)[
    [
        "Final_Priority_Rank",
        "Depot_ID",
        "Depot_Name",
        "District",
        "Terrain_Class",
        "Avg_EV_Suitability",
        "Avg_Operational_Intensity",
        "Avg_Passenger_Demand",
        "Avg_Passengers",
        "Avg_Effective_KM",
        "Total_OPEX_Saving_INR",
        "Total_CO2_Tonnes",
        "ML_Dominant_Category",
        "EV_Suitable_Pct",
        "EV_Transition_Priority_Score"
    ]
]

display(top10)
```

```python
# ============================================================
# FINAL TOP-10 ML VALIDATION
# ============================================================

# Calculate average ML prediction confidence for each depot
ml_confidence = (
    monthly_ml.groupby("Depot_ID")
    .agg(
        Avg_ML_Confidence=("Prediction_Confidence", "mean"),
        Min_ML_Confidence=("Prediction_Confidence", "min"),
        EV_Suitable_Months=(
            "Predicted_EV_Category",
            lambda x: (x == "EV Suitable").sum()
        ),
        Conditional_Months=(
            "Predicted_EV_Category",
            lambda x: (x == "Conditional").sum()
        ),
        Diesel_Preferred_Months=(
            "Predicted_EV_Category",
            lambda x: (x == "Diesel Preferred").sum()
        )
    )
    .reset_index()
)

# Merge ML validation results
top10_ml_test = top10.merge(
    ml_confidence,
    on="Depot_ID",
    how="left"
)

# Calculate percentage of months predicted EV Suitable
top10_ml_test["ML_EV_Suitable_Pct"] = (
    top10_ml_test["EV_Suitable_Months"] /
    (
        top10_ml_test["EV_Suitable_Months"]
        + top10_ml_test["Conditional_Months"]
        + top10_ml_test["Diesel_Preferred_Months"]
    )
) * 100

# Display validation
display(
    top10_ml_test[
        [
            "Final_Priority_Rank",
            "Depot_ID",
            "Depot_Name",
            "ML_Dominant_Category",
            "EV_Suitable_Pct",
            "ML_EV_Suitable_Pct",
            "Avg_ML_Confidence",
            "Min_ML_Confidence",
            "Avg_EV_Suitability",
            "EV_Transition_Priority_Score"
        ]
    ]
)
```

```python
# ============================================================
# FINAL DEPOT-LEVEL IMPACT ANALYSIS
# Correct annualization + potential CO2 avoidance
# ============================================================

import numpy as np
import pandas as pd

impact = depot_impact.copy()

# ------------------------------------------------------------
# 1. Calculate number of observed months for each depot
# ------------------------------------------------------------

observed_months = (
    monthly_ml
    .groupby("Depot_ID")
    .size()
    .reset_index(name="Observed_Months")
)

impact = impact.merge(
    observed_months,
    on="Depot_ID",
    how="left"
)

# ------------------------------------------------------------
# 2. Annualization factor
# ------------------------------------------------------------
# Annual value = observed total / observed months × 12

annual_factor = 12 / impact["Observed_Months"]

impact["Annual_Diesel_Litres"] = (
    impact["Total_Diesel_Litres"] * annual_factor
)

impact["Annual_CO2_Baseline_Tonnes"] = (
    impact["Total_CO2_Tonnes"] * annual_factor
)

impact["Annual_OPEX_Saving_INR"] = (
    impact["Total_OPEX_Saving_INR"] * annual_factor
)

impact["Annual_EV_Energy_MWh"] = (
    impact["Total_EV_Energy_MWh"] * annual_factor
)

# ------------------------------------------------------------
# 3. Potential CO2 avoided
# ------------------------------------------------------------
# IMPORTANT:
# This assumes 100% replacement of the corresponding
# diesel operation by zero-tailpipe-emission EVs.
#
# It is a POTENTIAL avoidance estimate, not an actual
# measured future reduction.

impact["Potential_CO2_Avoided_100pct_Tonnes"] = (
    impact["Annual_CO2_Baseline_Tonnes"]
)

# ------------------------------------------------------------
# 4. Example 25% transition scenario
# ------------------------------------------------------------
# Only 25% of diesel operation is assumed to transition.

impact["Potential_CO2_Avoided_25pct_Tonnes"] = (
    impact["Annual_CO2_Baseline_Tonnes"] * 0.25
)

impact["Potential_OPEX_Saving_25pct_INR"] = (
    impact["Annual_OPEX_Saving_INR"] * 0.25
)

# ------------------------------------------------------------
# 5. Display final impact table
# ------------------------------------------------------------

impact_columns = [
    "Depot_ID",
    "Depot_Name",
    "District",
    "Observed_Months",
    "Annual_Diesel_Litres",
    "Annual_CO2_Baseline_Tonnes",
    "Potential_CO2_Avoided_100pct_Tonnes",
    "Potential_CO2_Avoided_25pct_Tonnes",
    "Annual_OPEX_Saving_INR",
    "Potential_OPEX_Saving_25pct_INR",
    "Annual_EV_Energy_MWh"
]

final_impact = (
    impact[impact_columns]
    .sort_values("Annual_OPEX_Saving_INR", ascending=False)
    .reset_index(drop=True)
)

print("=" * 80)
print("FINAL 92-DEPOT IMPACT ANALYSIS")
print("=" * 80)

print("\nNumber of depots:", final_impact["Depot_ID"].nunique())

display(final_impact.head(10))
```

```python
# ============================================================
# FUTURE SCENARIO PREDICTION
# New Decision Tree Model
# ============================================================

import joblib
import numpy as np
import pandas as pd

# ------------------------------------------------------------
# Load saved model
# ------------------------------------------------------------

scenario_model = joblib.load(
    "../models/tejas_ev_priority_decision_tree.pkl"
)

# ------------------------------------------------------------
# PROJECT ASSUMPTIONS
# Taken from the dataset
# ------------------------------------------------------------

DIESEL_EFFICIENCY = 4.08       # km/litre
DIESEL_COST = 51.0             # INR/km
EV_COST = 27.0                 # INR/km
EV_ENERGY_RATE = 1.25          # kWh/km
CO2_PER_LITRE = 0.00268        # tonnes CO2/litre

TERRAIN_SCORES = {
    "Flat": 1.00,
    "Flat/Rolling": 0.90,
    "Rolling": 0.75,
    "Hilly": 0.40,
    "Steep": 0.20
}


# ============================================================
# SCENARIO FUNCTION
# ============================================================

def predict_future_scenario(
    buses,
    schedules,
    passengers,
    effective_km,
    terrain_class
):

    # --------------------------------------------------------
    # Validate terrain
    # --------------------------------------------------------

    if terrain_class not in TERRAIN_SCORES:
        raise ValueError(
            f"Invalid terrain. Choose from: {list(TERRAIN_SCORES.keys())}"
        )

    terrain_score = TERRAIN_SCORES[terrain_class]

    # --------------------------------------------------------
    # Derived diesel / environmental features
    # --------------------------------------------------------

    diesel_litres = (
        effective_km / DIESEL_EFFICIENCY
    )

    co2_tonnes = (
        diesel_litres * CO2_PER_LITRE
    )

    # --------------------------------------------------------
    # Derived EV energy
    # --------------------------------------------------------

    ev_energy_mwh = (
        effective_km * EV_ENERGY_RATE / 1000
    )

    # --------------------------------------------------------
    # Derived economic impact
    # --------------------------------------------------------

    opex_saving = (
        effective_km * (DIESEL_COST - EV_COST)
    )

    # --------------------------------------------------------
    # Derived operational ratios
    # --------------------------------------------------------

    passengers_per_bus = (
        passengers / buses
        if buses > 0 else 0
    )

    passengers_per_schedule = (
        passengers / schedules
        if schedules > 0 else 0
    )

    diesel_litres_per_bus = (
        diesel_litres / buses
        if buses > 0 else 0
    )

    diesel_litres_per_schedule = (
        diesel_litres / schedules
        if schedules > 0 else 0
    )

    co2_per_bus = (
        co2_tonnes / buses
        if buses > 0 else 0
    )

    # --------------------------------------------------------
    # Construct EXACT 13-feature vector
    # --------------------------------------------------------

    scenario = pd.DataFrame([{
        "Buses Allocated": buses,
        "Schedules Allocated": schedules,
        "Passengers": passengers,
        "Estimated Diesel Litres": diesel_litres,
        "Estimated CO2 (Tonnes)": co2_tonnes,
        "Effective KM": effective_km,
        "Estimated EV Energy (MWh)": ev_energy_mwh,
        "Passengers_per_Bus": passengers_per_bus,
        "Passengers_per_Schedule": passengers_per_schedule,
        "Diesel_Litres_per_Bus": diesel_litres_per_bus,
        "Diesel_Litres_per_Schedule": diesel_litres_per_schedule,
        "CO2_per_Bus_Tonnes": co2_per_bus,
        "Terrain_Score": terrain_score
    }])

    # --------------------------------------------------------
    # ML prediction
    # --------------------------------------------------------

    prediction = scenario_model.predict(scenario)[0]

    probabilities = scenario_model.predict_proba(scenario)[0]

    probability_df = pd.DataFrame({
        "Category": scenario_model.classes_,
        "Probability": probabilities * 100
    }).sort_values(
        "Probability",
        ascending=False
    ).reset_index(drop=True)

    confidence = probability_df.iloc[0]["Probability"]

    # --------------------------------------------------------
    # Display
    # --------------------------------------------------------

    print("=" * 75)
    print("FUTURE SCENARIO EV TRANSITION PREDICTION")
    print("=" * 75)

    print("\nUSER SCENARIO")
    print("-" * 45)
    print("Buses             :", buses)
    print("Schedules         :", schedules)
    print("Passengers        :", f"{passengers:,.0f}")
    print("Effective KM      :", f"{effective_km:,.0f}")
    print("Terrain           :", terrain_class)
    print("Terrain Score     :", terrain_score)

    print("\nDERIVED IMPACT METRICS")
    print("-" * 45)
    print("Diesel Litres     :", f"{diesel_litres:,.2f}")
    print("CO2 Baseline      :", f"{co2_tonnes:,.2f} tonnes")
    print("EV Energy         :", f"{ev_energy_mwh:,.2f} MWh")
    print("OPEX Saving       :", f"₹{opex_saving:,.2f}")

    print("\nML PREDICTION")
    print("-" * 45)
    print("Prediction        :", prediction)
    print("Confidence        :", f"{confidence:.2f}%")

    print("\nCLASS PROBABILITIES")
    print("-" * 45)

    display(
        probability_df.round(2)
    )

    return {
        "prediction": prediction,
        "confidence": confidence,
        "probabilities": probability_df,
        "features": scenario,
        "diesel_litres": diesel_litres,
        "co2_tonnes": co2_tonnes,
        "ev_energy_mwh": ev_energy_mwh,
        "opex_saving": opex_saving
    }
```
